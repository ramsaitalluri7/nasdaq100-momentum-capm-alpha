```python
def screen_ticker_completeness(tickers_close, rename_by_year):
    """
    逐年检查每个应有ticker的数据完整度。
    只打印数据不完整的ticker：年份 - ticker - 有效天数/该年总天数
    """
    columns_set = set(tickers_close.columns)
    zero_list = []
    new_ticker_year = {}
    for year, expected_tickers in rename_by_year.items():
        eligible_tickers = list(columns_set & set(expected_tickers))
        year_data = tickers_close.loc[lambda d: d.index.year == year, eligible_tickers]
        total_days = len(year_data)   # 该年份实际的交易日总数（以index行数为准）

        # 每支ticker在这一年里，非NaN的天数
        valid_counts = year_data.notna().sum()   # Series: index=ticker, value=有效天数

        instrument_set_eligible = set(eligible_tickers)
        for ticker, valid_days in valid_counts.items():
            if valid_days < total_days:
                print(f"{year} - {ticker}: {valid_days}/{total_days}")
                if valid_days == 0:
                    zero_list.append(ticker)
                    instrument_set_eligible-= {ticker}

        new_ticker_year[year] = list(instrument_set_eligible)
    return zero_list, new_ticker_year

print(f"existing imcomplete columns, and only zero should be deleted\n")
zero_list, eligible_tickers = screen_ticker_completeness(tickers_close, rename_by_year)

## "zero_list" is the drop list, "eligible_tickers" is the rest list with group
```

```python
def simple_return(price):
    """
    from price df to simple return df
    """
    out=\
    (
        price
        .pct_change()
        .fillna(0)
    )
    return out

def return_curve(returns):
    """
    from simple return df to cum_return df
    """
    out=\
    (
        returns
        .add(1)
        .cumprod()
    )
    return out
```

```python
### strategy1：common long-short momentum portfolio
def labels_long_short_mean_return(simple_return, q):
    """
    from simple return df to get the labels_list with different sorted groups
    """
    labels = ["short"]
    labels.extend([i for i in range(q)[2:]])
    labels.append("long")
    labels_dict=\
    (
        simple_return
        .mean(axis = 0)
        .sort_values(ascending = False)
        .pipe(pd.qcut,
              q,
              labels = labels,
              duplicates="drop"
        )
        .to_frame()
        .groupby(0)
        .groups           # 将分组结果，以字典形式输出
    )
    return labels_dict

def long_short_return(simple_return, labels_dict):
    long =\
    (
        simple_return
        [labels_dict['long']]
        .apply(return_curve)
        .iloc[-1, :]
        .mean()
    )
    short =\
    (
        simple_return
        [labels_dict['short']]
        .apply(return_curve)
        .iloc[-1, :]
        .mean()
    )
    return long - short
```

### Section 3. Rolling with given window and backtest

```python
### Rolling backtest — 
### filters to (a) point-in-time index members
### AND (b) tickers with real (non-all-NaN) price data in the fit window

def rolling_backtest(assets_simple_return, eligible_tickers,
                      fit_window=20, trade_window=5, n_quantile=5):
    
    records_trade_events = []
    records_return = []

    years_available = sorted(eligible_tickers.keys())
    k = 0
    n = assets_simple_return.shape[0]

    while fit_window + k * trade_window < n:
        fit_start = k * trade_window
        fit_end   = fit_start + fit_window
        trade_start = fit_end
        trade_end   = min(trade_start + trade_window, n)

        # --- 用fit窗口最后一天的日期，作为信号的"as-of"锚点 ---
        as_of_date = assets_simple_return.index[fit_end - 1]
        year = as_of_date.year

        # --- 年份边界兜底：早于/晚于覆盖范围，用最近的边界年份 ---
        if year < years_available[0]:
            year = years_available[0]
        elif year > years_available[-1]:
            year = years_available[-1]

        valid_tickers = eligible_tickers[year]

        # --- (a) point-in-time成分股 + 真实存在于数据列中 ---
        eligible_cols = [c for c in assets_simple_return.columns if c in valid_tickers]

        window_data = assets_simple_return.iloc[fit_start:fit_end][eligible_cols]

        # --- (b) 排除fit窗口内全为0/NaN（未上市/已退市）的ticker ---
        tradable_cols = window_data.columns[(window_data != 0).any(axis=0)]
        window_data = window_data[tradable_cols]

        # 资产数不足以分组时，跳过这一期，避免qcut报错或组别为空
        if len(window_data.columns) < n_quantile:
            print(f"警告: {as_of_date.date()} 可用资产数({len(window_data.columns)}) < n_quantile({n_quantile})，跳过该期")
            k += 1
            continue

        # --- 1) 只在每个再平衡点fit一次信号 ---
        screenshot_labels_dict = (
            window_data
            .pipe(labels_long_short_mean_return, n_quantile)
        )

        long_list  = list(screenshot_labels_dict.get("long", []))
        short_list = list(screenshot_labels_dict.get("short", []))

        records_trade_events.append([as_of_date, long_list, short_list])

        # --- 2) 用同一套信号，持有trade_window周，逐周记录收益 ---
        for t in range(trade_start, trade_end):
            weekly_trade_return = (
                assets_simple_return
                .iloc[t: t+1, :]
                .pipe(long_short_return, screenshot_labels_dict)
            )

            records_return.append([
                assets_simple_return.index[t],
                weekly_trade_return
            ])

        k += 1

    return records_trade_events, records_return

def backtest_metrics(returns, periods_per_year=52):
    final_multiple = (returns + 1).cumprod().iloc[-1]
    span = (returns.index[-1] - returns.index[0]).days
    CAGR = final_multiple**(365/span) - 1

    mean = returns.mean()
    std = returns.std()
    Sharpe = mean/std * np.sqrt(periods_per_year)

    equity_curve = (returns + 1).cumprod()
    MDD = equity_curve.div(equity_curve.cummax()).sub(1).min()

    std_ann = std * np.sqrt(periods_per_year)

    return final_multiple, CAGR, Sharpe, MDD, std_ann

def draw_down_series(simple_return):
    cum_return =\
    (
        return_curve(simple_return)
    )

    drawdown_series =\
    (
        cum_return
        .cummax()
        .pipe(cum_return.div)
        .sub(1)
    )
    return drawdown_series

def longest_drawdown_run(simple_return):
    return_curve =\
    (
        simple_return
        .add(1)
        .cumprod()
    )
    underwater = return_curve < return_curve.cummax()  # True is underwater

    max_count = 0
    count = 0
    best_end = None
    for date, isunderwater in underwater.items():
        if isunderwater:
            count+=1
            if count > max_count:
                best_end = date
                max_count = count
        else:
            count = 0

    best_start = underwater.index[underwater.index.get_loc(best_end) - max_count + 1]

    if best_end == underwater.index[-1]:
        print("Keep underwaters...")

    return max_count, str(best_start)[:-9], str(best_end)[:-9]
```

### Section 5. Robustness test

```python
def matrixs(assets_simple_return, index_simple_return, eligible_tickers, n_quantile = 10,
            fit_step=range(1, 30, 3), trade_step=range(1, 10)):
    
    SR_matrixs = pd.DataFrame(
        np.zeros([len(fit_step), len(trade_step)]),
        index=fit_step, columns=trade_step
    )
    alpha_matrixs = SR_matrixs.copy()
    p_alpha_matrxs = SR_matrixs.copy()

    for i in fit_step:
        for j in trade_step:
            _, records_return = rolling_backtest(
                assets_simple_return.iloc[1:],
                eligible_tickers,
                i, j, n_quantile
            )

            strategy_simple_return = (
                pd.DataFrame(records_return, columns=["Date", "Strategy"])
                .set_index("Date")
            )

            rolling_simple_return = (
                pd.concat([strategy_simple_return, index_simple_return], axis=1, join="inner")
                .rename(columns={"^NDX": "Benchmark"})
            )

            SR_diff = (
                rolling_simple_return
                .apply(backtest_metrics, periods_per_year=52)
                .assign(Ratio="final_multiple,  CAGR, Sharpe, MDD, std_ann".split(", "))
                .set_index("Ratio")
                .iloc[2, :]
                .diff()
                .mul(-1)
                .values[-1]
            )

            x = rolling_simple_return.Benchmark
            y = rolling_simple_return.Strategy
            result = stats.linregress(x, y)

            t_alpha = result.intercept / result.intercept_stderr
            p_alpha_alpha = 2 * (1 - stats.t.cdf(abs(t_alpha), df=len(x) - 2))

            SR_matrixs.loc[i, j] = SR_diff
            alpha_matrixs.loc[i, j] = result.intercept
            p_alpha_matrxs.loc[i, j] = p_alpha_alpha

    return SR_matrixs, alpha_matrixs, p_alpha_matrxs
```

```python
def matrix_to_long(matrix, value_name, period_label):
    return (
        matrix
        .rename_axis("fit_window")
        .reset_index()
        .melt(id_vars="fit_window", var_name="trade_window", value_name=value_name)
        .assign(period=period_label)
    )
```
